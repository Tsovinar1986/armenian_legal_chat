# Legal UI — live preview from a notebook

Embeds the real Vite frontend (`frontend/`) talking to the real FastAPI backend (`api.py`) in an `IFrame`, so you can see it working without leaving Jupyter. This is the same page as `http://localhost:5173` — nothing here is a separate copy or a mock.

**Before running the cells below, in two terminals from the repo root:**

```bash
uvicorn api:app --reload --host 0.0.0.0 --port 8000   # backend
cd frontend && npm install && npm run dev              # frontend (localhost:5173)
```

````python
import requests

def check(url, name):
    try:
        r = requests.get(url, timeout=2)
        print(f"\u2705 {name} is up ({url}) \u2014 HTTP {r.status_code}")
        return True
    except requests.exceptions.RequestException as exc:
        print(f"\u274c {name} is not reachable at {url} ({exc.__class__.__name__}). Start it first \u2014 see the markdown cell above.")
        return False

backend_up = check("http://localhost:8000/health", "Backend (uvicorn)")
frontend_up = check("http://localhost:5173", "Frontend (vite)")
````

````python
from IPython.display import IFrame

# Live, interactive — type a question, upload a file, or use the mic button
# right here. It's making real requests to localhost:8000 as you use it.
IFrame(src="http://localhost:5173", width="100%", height=800)
````
